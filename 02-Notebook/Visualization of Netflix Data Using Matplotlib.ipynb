```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("Cleaned_Netflix_Data.csv")
```

```python
#Type of Films
#pie chart
size=df["type"].value_counts()
plt.pie(size,labels=size.index,autopct="%1.1f%%")
plt.title("Types of Films")
plt.show()
```

```python
#Films with Highest Runtime
Type=df["type"].value_counts()
df.loc[df.groupby("type")["runtime"].idxmax()]
```

```python
#Year of Highest Votes Listed
#Line Plot
df["imdb_votes"] = (df["imdb_votes"].astype(str).str.replace(",", "", regex=False))
df["imdb_votes"] = pd.to_numeric(df["imdb_votes"], errors="coerce")
votes_per_year = df.groupby("release_year")["imdb_votes"].max()

plt.figure(figsize=(12,6))

plt.plot(votes_per_year.index,votes_per_year.values,marker="o")

plt.xlabel("Release Year")
plt.ylabel("Maximum IMDb Votes")
plt.title("Highest IMDb Votes by Release Year")

plt.grid(True)

plt.show()
```

```python
# Film releases after 2015
#Line Plot
before_2015 = len(df[df["release_year"] < 2015])
after_2015 = len(df[df["release_year"] >= 2015])

if before_2015 > after_2015:
    print(
        "There are more films released before 2015.","Total", before_2015)
else:
    print("There are more films released after 2015.","Total", after_2015)

# Line plot
x = ["Before 2015", "2015 & After"]
y = [before_2015, after_2015]

plt.figure(figsize=(8,5))

plt.plot(x, y, marker="o", linewidth=3, color="green")

plt.xlabel("Time Period")
plt.ylabel("Number of Films")
plt.title("Film Releases After 2015")

plt.grid(True)
plt.show()
```

```python
#Count of Year with Highest Film Releases
#Donut Chart
donut=df["release_year"].sort_index(ascending=False).value_counts().head()
plt.pie(donut,labels=donut.index,autopct="%1.1f%%",wedgeprops={'width':0.4})
plt.title("Count of Year with Highest Film Releases")
plt.show()
```

```python
#Films By Highest imdb Scores
#Bar Chart
bar_data=df.groupby("title")["imdb_score"].max().sort_values(ascending=False).head()
plt.figure(figsize=(10,6))
plt.bar(bar_data.index,bar_data.values)
plt.title("Films By Highest imdb Scores")
plt.xlabel("Films")
plt.ylabel("imdb Scores")
plt.show()
```

```python
#Films By Highest imdb Votes
#Area Chart
line=df.groupby("title")["imdb_votes"].max().sort_values(ascending=False).head()
plt.figure(figsize=(15,5))
plt.title("Films By Highest imdb Votes")
plt.xlabel("Films")
plt.ylabel("imdb Votes")
plt.plot(line.index,line.values,color="red",marker="o")
plt.fill_between(line.index,line.values,color="skyblue",alpha=0.5)
plt.show()
```

```python
#Sum of films By Age Certification
#Horizontal Bar Chart
barh_data=df["age_certification"].value_counts()
plt.figure(figsize=(12,6))
plt.barh(barh_data.index,barh_data.values)
plt.title("Sum of films By Age Certification")
plt.xlabel("Number of films")
plt.ylabel("Age Certifications")
plt.show()
```
