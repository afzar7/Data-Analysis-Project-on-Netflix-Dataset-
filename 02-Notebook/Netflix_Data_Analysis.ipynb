```python
#Loading Dataset
import pandas as pd
dF=pd.read_csv("Raw_Netflix_Data.csv")
print(dF.head())
```

```python
#Viewing Columns in dataset
dF.columns
```

```python
#Counting NULL Values Before Cleaning
dF.isnull().sum()
```

```python
#removing_duplicates
df=dF.drop_duplicates()
```

```python
#Filling NULL Values 
df["title"] = df["title"].str.lstrip("#")
df['age_certification'] = df['age_certification'].fillna("Unrated")
df['description'] = df['description'].fillna("Unknown")
df['imdb_votes'] = df['imdb_votes'].fillna(1)
```

```python
#Counting NULL Values After Cleaning
df.isnull().sum()
```

```python
#Type of Films
df["type"].value_counts()
```

```python
#Finding Movie & Show With Highest Runtime
Type=df["type"].value_counts()
df.loc[df.groupby("type")["runtime"].idxmax()]
```

```python
#Year of Highest Votes Listed
df["imdb_votes"] = (df["imdb_votes"].astype(str).str.replace(",", "", regex=False))
df["imdb_votes"] = pd.to_numeric(df["imdb_votes"],errors="coerce")
max_votes_row = df.loc[df["imdb_votes"].idxmax()]
year = max_votes_row["release_year"]
max_votes = max_votes_row["imdb_votes"]
print("YEAR:", year, "- Maximum Votes:", max_votes)
```

```python
#Film releases after 2015
before_2015=len(df[df["release_year"]<2015])
after_2015=len(df[df["release_year"]>=2015])
if before_2015 > after_2015:
    print("There are more movies released before 2015,","Total",before_2015,"Films released before 2015")
else:
    print("There are more movies released after 2015,","Total",after_2015,"Films released after 2015")
```

```python
#Count of Year with Highest Film Releases
df["release_year"].sort_index(ascending=False).value_counts().head()
```

```python
#Films By Highest imdb Scores
df.groupby("title")["imdb_score"].max().sort_values(ascending=False).head()
```

```python
#Films By Highest imdb Votes
df.groupby("title")["imdb_votes"].max().sort_values(ascending=False).head()
```

```python
#Films By Age Certification
df["age_certification"].value_counts(ascending=False)
```

```python
#Saving Cleaned Data
df.to_csv("cleaned_netflix_data.csv", index=False)
```
